````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
print(calculator("20 + 5"))
print(calculator("100 / 4"))
print(calculator("10 * 8"))
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

````python
text = "Artificial Intelligence is transforming industries"

print(extract_keywords(text))
````

````python
# 🤖 AGENT FUNCTION

def agent(query: str):

    try:

        # Check if query is valid
        if not isinstance(query, str) or not query.strip():
            return {
                "type": "error",
                "result": "Invalid or empty query"
            }

        query_lower = query.lower()

        # Route to calculator
        if "calculate" in query_lower:
            expression = query_lower.replace("calculate", "").strip()

            if not expression:
                return {
                    "type": "error",
                    "result": "No mathematical expression provided"
                }

            result = calculator(expression)

            if result == "Error in calculation":
                return {
                    "type": "error",
                    "result": result
                }

            return {
                "type": "calculation",
                "result": result
            }

        # Route to keyword extractor
        elif "keywords" in query_lower:
            text = (
                query_lower
                .replace("keywords", "")
                .replace("extract", "")
                .replace("from", "")
                .strip()
            )

            if not text:
                return {
                    "type": "error",
                    "result": "No text provided for keyword extraction"
                }

            result = extract_keywords(text)

            return {
                "type": "keywords",
                "result": result
            }

        # General query
        else:
            return {
                "type": "general",
                "result": query
            }

    except Exception as e:
        return {
            "type": "error",
            "result": str(e)
        }
````

````python
# 🚀 BONUS: Improved Routing

def improved_agent(query: str):

    try:

        # Check if query is valid
        if not isinstance(query, str) or not query.strip():
            return {
                "type": "error",
                "result": "Invalid or empty query"
            }

        query_lower = query.lower()

        # --------------------------------
        # ROUTE 1: IMPROVED CALCULATOR
        # --------------------------------

        calculation_words = [
            "calculate",
            "compute",
            "solve"
        ]

        if any(word in query_lower for word in calculation_words):

            expression = query_lower

            for word in calculation_words:
                expression = expression.replace(word, "")

            expression = expression.replace("what is", "")
            expression = expression.replace("?", "")
            expression = expression.strip()

            if not expression:
                return {
                    "type": "error",
                    "result": "No mathematical expression provided"
                }

            result = calculator(expression)

            if result == "Error in calculation":
                return {
                    "type": "error",
                    "result": result
                }

            return {
                "type": "calculation",
                "result": result
            }

        # --------------------------------
        # ROUTE 2: IMPROVED KEYWORD TOOL
        # --------------------------------

        keyword_words = [
            "keywords",
            "keyword"
        ]

        if any(word in query_lower for word in keyword_words):

            text = query_lower

            text = text.replace("extract keywords", "")
            text = text.replace("find keywords", "")
            text = text.replace("keywords", "")
            text = text.replace("keyword", "")
            text = text.replace("extract", "")
            text = text.replace("from", "")

            text = text.strip()

            if not text:
                return {
                    "type": "error",
                    "result": "No text provided for keyword extraction"
                }

            result = extract_keywords(text)

            return {
                "type": "keywords",
                "result": result
            }

        # --------------------------------
        # ROUTE 3: GENERAL QUERY
        # --------------------------------

        return {
            "type": "general",
            "result": query
        }

    except Exception as e:

        return {
            "type": "error",
            "result": str(e)
        }
````

````python
# 📝 BONUS: Logging Setup

import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logger = logging.getLogger("SmartAssistant")
````

````python
# 📝 BONUS: Agent With Logging

def logged_agent(query: str):

    logger.info(f"Query received: {query}")

    try:

        # Check input
        if not isinstance(query, str) or not query.strip():

            logger.error("Invalid or empty query")

            return {
                "type": "error",
                "result": "Invalid or empty query"
            }

        query_lower = query.lower()

        # --------------------------------
        # CALCULATOR ROUTE
        # --------------------------------

        if "calculate" in query_lower:

            logger.info("Route selected: calculator")

            expression = query_lower.replace("calculate", "").strip()

            if not expression:

                logger.error("No mathematical expression provided")

                return {
                    "type": "error",
                    "result": "No mathematical expression provided"
                }

            logger.info(f"Calculator input: {expression}")

            result = calculator(expression)

            logger.info(f"Calculator result: {result}")

            if result == "Error in calculation":

                logger.error("Calculator failed")

                return {
                    "type": "error",
                    "result": result
                }

            return {
                "type": "calculation",
                "result": result
            }

        # --------------------------------
        # KEYWORD ROUTE
        # --------------------------------

        elif "keywords" in query_lower:

            logger.info("Route selected: keyword extractor")

            text = (
                query_lower
                .replace("keywords", "")
                .replace("extract", "")
                .replace("from", "")
                .strip()
            )

            if not text:

                logger.error("No text provided for keyword extraction")

                return {
                    "type": "error",
                    "result": "No text provided for keyword extraction"
                }

            logger.info(f"Keyword extraction input: {text}")

            result = extract_keywords(text)

            logger.info(f"Extracted keywords: {result}")

            return {
                "type": "keywords",
                "result": result
            }

        # --------------------------------
        # GENERAL ROUTE
        # --------------------------------

        else:

            logger.info("Route selected: general response")

            return {
                "type": "general",
                "result": query
            }

    except Exception as e:

        logger.exception("Unexpected error occurred")

        return {
            "type": "error",
            "result": str(e)
        }
````

````python
# 🛠️ BONUS TOOL 3: Word Counter

def word_counter(text: str) -> int:
    """Count the number of words in a text."""

    try:
        words = text.split()
        return len(words)

    except Exception:
        return 0
````

# 🏆 FINAL INTEGRATED AGENT

````python
# 🏆 STEP 1: FINAL INTEGRATED AGENT

def final_agent(query: str):

    logger.info(f"Query received: {query}")

    try:

        # --------------------------------
        # 1. INPUT VALIDATION
        # --------------------------------

        if not isinstance(query, str) or not query.strip():

            logger.error("Invalid or empty query")

            return {
                "type": "error",
                "result": "Invalid or empty query"
            }

        query_lower = query.lower().strip()

        # --------------------------------
        # 2. CALCULATOR ROUTING
        # --------------------------------

        calculation_words = [
            "calculate",
            "compute",
            "solve"
        ]

        if any(word in query_lower for word in calculation_words):

            logger.info("Route selected: calculator")

            expression = query_lower

            for word in calculation_words:
                expression = expression.replace(word, "")

            expression = expression.replace("what is", "")
            expression = expression.replace("?", "")
            expression = expression.strip()

            if not expression:

                logger.error("No mathematical expression provided")

                return {
                    "type": "error",
                    "result": "No mathematical expression provided"
                }

            logger.info(f"Calculator input: {expression}")

            result = calculator(expression)

            logger.info(f"Calculator result: {result}")

            if result == "Error in calculation":

                logger.error("Calculator failed")

                return {
                    "type": "error",
                    "result": result
                }

            return {
                "type": "calculation",
                "result": result
            }

        # --------------------------------
        # 3. KEYWORD ROUTING
        # --------------------------------

        keyword_words = [
            "keywords",
            "keyword"
        ]

        if any(word in query_lower for word in keyword_words):

            logger.info("Route selected: keyword extractor")

            text = query_lower

            text = text.replace("extract keywords", "")
            text = text.replace("find keywords", "")
            text = text.replace("keywords", "")
            text = text.replace("keyword", "")
            text = text.replace("extract", "")
            text = text.replace("from", "")

            text = text.strip()

            if not text:

                logger.error("No text provided for keyword extraction")

                return {
                    "type": "error",
                    "result": "No text provided for keyword extraction"
                }

            logger.info(f"Keyword extraction input: {text}")

            result = extract_keywords(text)

            logger.info(f"Extracted keywords: {result}")

            return {
                "type": "keywords",
                "result": result
            }

        # --------------------------------
        # 4. WORD COUNT ROUTING
        # --------------------------------

        if "count words" in query_lower or "word count" in query_lower:

            logger.info("Route selected: word counter")

            if "count words" in query_lower:

                text = query_lower.split("count words", 1)[1]

            else:

                text = query_lower.split("word count", 1)[1]

            text = text.strip()

            if text.startswith("in "):

                text = text[3:].strip()

            if not text:

                logger.error("No text provided for word counting")

                return {
                    "type": "error",
                    "result": "No text provided for word counting"
                }

            logger.info(f"Word counter input: {text}")

            result = word_counter(text)

            logger.info(f"Word count result: {result}")

            return {
                "type": "word_count",
                "result": result
            }

        # --------------------------------
        # 5. GENERAL QUERY
        # --------------------------------

        logger.info("Route selected: general response")

        return {
            "type": "general",
            "result": query
        }

    # --------------------------------
    # 6. GLOBAL ERROR HANDLING
    # --------------------------------

    except Exception as e:

        logger.exception("Unexpected error occurred")

        return {
            "type": "error",
            "result": str(e)
        }
````
